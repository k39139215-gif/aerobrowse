### 2. Fetch Any Live Website (Sub-100ms & 88% Fewer Tokens)
Watch how AeroBrowse converts raw web HTML into dense, high-signal Markdown without losing visible text:

```python
from aerobrowse import AeroBrowser

browser = AeroBrowser()
result = browser.fetch("https://quotes.toscrape.com")

print(f"📌 Page Title:         {result['title']}")
print(f"📦 Raw HTML Size:      {result['original_chars']:,} characters")
print(f"🧠 Pruned Markdown:    {result['pruned_chars']:,} characters")
print(f"⚡ Token Reduction:    {result['reduction_pct']}% saved!")
print(f"🎯 Interactive Target: {len(result['actions'])} Set-of-Marks mapped")

print("\n--- First 300 Characters of Clean Markdown ---")
print(result["markdown"][:300])
```

### 3. Numbered Set-of-Marks Action Targets
Instead of hallucinating complex CSS selectors (`div > ul > li:nth-child(3)`), LLMs can simply reference clean numeric IDs:

```python
for action in result["actions"][:5]:
    print(f"[{action['id']}] {action['tag'].upper()}: '{action['label']}' -> {action['href']}")
```

### 4. Drop-in LangChain / CrewAI Integration
Plug AeroBrowse straight into your autonomous agent pipelines in 3 lines:

```python
# Drop-in tool for any agent framework
def browse_web_tool(url: str) -> str:
    """Fetches URL, extracts 100% human knowledge, saves 88% LLM tokens."""
    return browser.fetch(url)["markdown"]

output = browse_web_tool("https://books.toscrape.com")
print(f"Successfully retrieved {len(output)} chars of dense e-commerce catalog markdown!")
```
